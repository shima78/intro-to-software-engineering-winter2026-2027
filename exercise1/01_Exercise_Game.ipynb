**Note on the game loop:** each round runs in `play()`. After a crash, `game_over()` waits for **SPACE/ENTER** (play again) or **ESC** (quit), and `run()` keeps starting new rounds until the player quits. `pygame.quit()` is only called once at the end of `run()`.
