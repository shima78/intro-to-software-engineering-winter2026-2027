# 2. Python-Grundlagen

## Syntax (Folie 8)
In Python wird Code durch Einrückung statt durch Klammern strukturiert.
Im Vergleich zu C++ oder Java müssen Variablen nicht mit einem bestimmten Typ initialisiert werden.

| C++ | Java |
|---|---|
| <pre>int x=4;<br>int y=7;<br>if (x > y){<br>std::cout << "x bigger than y";<br>}</pre> | <pre>int x=4;<br>int y=7;<br>if (x > y){<br>System.out.println("x bigger than y");<br>}</pre> |

````python
# 📄 Folie 8 – Python
x = 4
y = 7
if x > y:
    print("x bigger than y")
````

❓ **Warum gibt es keine Ausgabe?** `4 > 7` ist `False`, also wird der eingerückte Block übersprungen.
Ändere `x` auf `10` und führe die Zelle erneut aus.

## Funktionen (Folie 9)
Eine Funktion wird mit `def` deklariert. Sie kann Eingabewerte entgegennehmen und Werte zurückgeben.

````python
# 📄 Folie 9 – Beispielfunktion
def add(a, b):
    result = a + b
    return result
````

## Aufruf einer Funktion (Folie 10)

````python
# 📄 Folie 10 – Variablen übergeben
num1 = 3
num2 = 5
sum = add(num1 , num2)
````

````python
# ➕ Extra: Ergebnis anzeigen
print("sum  =", sum)
print("num1 =", num1, "| num2 =", num2, " <- nicht verändert")
````

````python
# 📄 Folie 10 – Werte übergeben
sum = add(3,5)
````

````python
# ➕ Extra
print("sum =", sum)
````

## Lokale Variablen (Folie 11)
Die Variablen `a`, `b` und `result` innerhalb der Funktion sind außerhalb der Funktion **nicht** bekannt.

*(Die Einrückung innerhalb von `add` wurde hier vereinheitlicht; auf der Folie sind die beiden Zeilen unterschiedlich eingerückt, was Python ablehnen würde.)*

````python
# 📄 Folie 11 – links
def add(a,b):
    a = a+b
    return a

a = 3
b = 5
sum = add(a,b)

#Welchen Wert hat a jetzt?
````

````python
# ➕ Extra: Auflösung
print("a   =", a)
print("sum =", sum)
````

````python
# 📄 Folie 11 – rechts
def add(a,b):
    a = a+b
    return a

a = 3
b = 5
a = add(a,b)

#Welchen Wert hat a jetzt?
````

````python
# ➕ Extra: Auflösung
print("a =", a)
````

- **Links:** `a` ist 3, weil die Funktion mit einer **lokalen Kopie** arbeitet und nicht mit der global definierten Variable `a`
- **Rechts:** `a` ist 8, da die Variable mit dem Rückgabewert der Funktion überschrieben wurde

### ⚠️ Stolperfalle: Wir haben gerade das eingebaute `sum()` überschrieben!
Die Folien 10 und 11 speichern das Ergebnis in einer Variable namens `sum`. Diese Variable verdeckt Pythons eingebaute Funktion `sum()`, die Folie 13 braucht.
Führe die nächste Zelle aus, um das Problem zu sehen, und danach die übernächste, um es zu beheben.

````python
# ➕ Extra: zeigt das Problem (löst absichtlich einen TypeError aus)
sum([1, 2, 3])
````

````python
# ➕ Extra: unsere Variable entfernen, damit das eingebaute sum() wieder sichtbar ist
del sum
print(sum([1, 2, 3]))
````

## Veränderlich (mutable) vs. unveränderlich (immutable) (Folie 12)
Änderungen an **unveränderlichen** Objekten innerhalb einer Funktion sind außerhalb nicht sichtbar. Änderungen an **veränderlichen** Objekten bleiben auch außerhalb der Funktion sichtbar.

| Veränderliche Typen | Unveränderliche Typen |
|---|---|
| Listen, Sets, Dictionaries | Zahlen (Integer, Rational, Float, Decimal, Complex & Booleans), Strings, Tupel |

## Folie 13

````python
# 📄 Folie 13 – links: übergebene Liste wird nicht verändert
def sum_values(numbers):
    total = sum(numbers)
    return total

my_numbers = [1, 2, 3, 4, 5]
result = sum_values(my_numbers)
print("Summe der Zahlen:", result)
# Ausgabe: Summe der Zahlen: 15
print("Ursprüngliche Liste:", my_numbers)
# Ausgabe: Ursprüngliche Liste: [1, 2, 3, 4, 5]
````

````python
# 📄 Folie 13 – rechts: übergebene Liste wird verändert
def sum_and_append(numbers):
    total = sum(numbers)
    numbers.append(total)
    return total

my_numbers = [1, 2, 3, 4, 5]
result = sum_and_append(my_numbers)
print("Summe der Zahlen:", result)
# Ausgabe: Summe der Zahlen: 15
print("Ursprüngliche Liste:", my_numbers)
# Ausgabe: Ursprüngliche Liste: [1, 2, 3, 4, 5, 15]
````

- **Links:** Die übergebene Variable `my_numbers` wird nicht verändert
- **Rechts:** Die übergebene Variable `my_numbers` wird mit `append` verändert, wodurch die berechnete Summe an die Liste angehängt wird. Die Funktion arbeitet **nicht** mit einer lokalen Kopie.

## Klassen (Folie 14)
Klassen werden mit `class` definiert. Die Attribute werden in `__init__()` mithilfe von `self` gesetzt.

````python
# 📄 Folie 14 – Klassendefinition
class mammal():
    def __init__(self, num_legs, fur, diet, territory):
        self.num_legs = num_legs
        self.fur = fur
        self.diet = diet
        self.territory = territory
````

````python
# 📄 Folie 14 – viele Instanzen mit unterschiedlichen Parametern
bear = mammal(4, True, "Omnivore", ["Foest"])
camel = mammal(4, True, "Herbivore", ["Desert"])
````

````python
# ➕ Extra: die beiden Instanzen ansehen
print("bear :", bear.num_legs, bear.fur, bear.diet, bear.territory)
print("camel:", camel.num_legs, camel.fur, camel.diet, camel.territory)
print(vars(bear))
````

## Vererbung (Folie 16)
Eine Kindklasse erbt Variablen und Methoden ihrer Elternklasse und kann neue hinzufügen.
`bear` erbt von der oben definierten Klasse `mammal` (4 Parameter, noch kein Hunger).

*(Auf der Folie ist die `super(...)`-Zeile aus `__init__` herausgerückt. Hier steht sie innerhalb von `__init__`, wo sie hingehört.)*

````python
# 📄 Folie 16
class bear(mammal):   # Das ist nur Pseudo-Code, um die Funktionsweise zu demonstrieren.
    def __init__(self):
        self.num_legs = 4
        self.fur = True
        self.diet = "Omnivore"
        self.territory = ["Forest"]
        super(bear, self).__init__(self.num_legs, self.fur, self.diet, self.territory)

    def roar(self):
        self.makes_noise()
````

````python
# ➕ Extra: ein bear wird erzeugt, ohne Werte zu übergeben
b = bear()
print(vars(b))
print("isinstance(b, mammal):", isinstance(b, mammal))
````

`roar()` ruft `makes_noise()` auf, das nirgends definiert ist (Pseudo-Code); `b.roar()` würde daher einen `AttributeError` auslösen.

## Hunger & Methoden (Folie 15)
Methoden sind Funktionen, die innerhalb einer Klasse definiert sind. Sie arbeiten meist mit den eigenen Variablen der Klasse.
Nun wird `mammal` um `hunger` und eine Methode `eat()` erweitert.

````python
# 📄 Folie 15
class mammal():   # Das ist nur Pseudo-Code, um die Funktionsweise zu demonstrieren.
    def __init__(self, num_legs, fur, diet, territory, hunger):
        self.num_legs = num_legs
        self.fur = fur
        self.diet = diet
        self.territory = territory
        self.hunger = hunger

    def eat(self, food, food_condition):
        if (food_condition is not "rotten" and self.is_compatible_with_diet(food)):
            self.hunger = 0
        else:
            self.vomit()
````

💡 Python ≥ 3.8 gibt für `is not "rotten"` eine `SyntaxWarning` aus: `is` prüft, ob zwei Namen auf dasselbe *Objekt* verweisen, nicht, ob die Werte *gleich* sind. Zum Vergleichen von Werten nimmt man `!=`.
Es ist **Pseudo-Code**: `is_compatible_with_diet()` und `vomit()` existieren noch nicht, daher schlägt der Aufruf von `eat()` fehl:

````python
# ➕ Extra: ausprobieren (löst absichtlich einen AttributeError aus)
m = mammal(4, True, "Omnivore", ["Forest"], hunger=10)
m.eat("berries", "fresh")
````

Hinweis: Die Klasse `bear` oben erbt noch vom *alten* `mammal` (ohne Hunger). Eine Neudefinition von `mammal` aktualisiert Klassen, die bereits davon abgeleitet wurden, nicht. Die vollständige Version unten definiert beide neu.

### ➕ Extra: der Pseudo-Code lauffähig gemacht
Dieselbe Idee wie in den Folien 15 + 16, aber mit den fehlenden Teilen ergänzt, damit wir die Methoden und die Vererbung tatsächlich in Aktion sehen.

````python
class mammal():
    def __init__(self, num_legs, fur, diet, territory, hunger=10):
        self.num_legs = num_legs
        self.fur = fur
        self.diet = diet
        self.territory = territory
        self.hunger = hunger

    def is_compatible_with_diet(self, food):
        plants = ["berries", "grass", "honey"]
        if self.diet == "Herbivore":
            return food in plants
        if self.diet == "Carnivore":
            return food not in plants
        return True  # Allesfresser (Omnivore) frisst alles

    def vomit(self):
        print("  Bäääh! 🤢")

    def makes_noise(self):
        print("  *irgendein Geräusch*")

    def eat(self, food, food_condition):
        print(f"{type(self).__name__} frisst {food_condition} {food} (Hunger vorher: {self.hunger})")
        if food_condition != "rotten" and self.is_compatible_with_diet(food):
            self.hunger = 0
        else:
            self.vomit()
        print(f"  Hunger nachher: {self.hunger}")


class bear(mammal):
    def __init__(self):
        super(bear, self).__init__(4, True, "Omnivore", ["Forest"])

    def roar(self):
        self.makes_noise()

    # Ein Kind kann auch eine Methode der Elternklasse überschreiben
    def makes_noise(self):
        print("  ROOOAAAR! 🐻")


camel = mammal(4, True, "Herbivore", ["Desert"])
teddy = bear()

camel.eat("grass", "fresh")    # funktioniert
camel.eat("fish", "fresh")     # passt nicht zur Ernährung
teddy.eat("fish", "rotten")    # verdorben
teddy.eat("honey", "fresh")    # eat() wird von mammal geerbt
teddy.roar()                   # roar() existiert nur in bear
# camel.roar()                 # -> AttributeError: mammal hat kein roar()
print("isinstance(teddy, mammal):", isinstance(teddy, mammal))
````
